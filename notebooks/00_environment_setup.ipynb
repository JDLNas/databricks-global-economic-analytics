```python
dados = [("Brasil", 2024), ("Argentina", 2024), ("Chile", 2024)]

df_paises = spark.createDataFrame(dados, ["pais", "ano"])

display(df_paises)
```

```python
df_paises.printSchema()
```

```python
df_paises.createOrReplaceTempView("vw_paises")
```

```python
resultado_sql = spark.sql("""
    SELECT
        pais,
        ano
    FROM vw_paises
    ORDER BY pais
""")

display(resultado_sql)
```
